# Exogenous Alpha

### Defining scenarios

With the @scenario macro you can easily set different parameters, the name must be the same as the model defined via the @model macro

```julia
scen1a = Static.@scenario Static.ExoAlpha begin
    γ = 0.8 # 0.5
end

scen2a = Static.@scenario Static.CBsense begin
    γ = 0.8 # 0.5
end

scen3a = Static.@scenario Static.DualInterest begin
    γ = 0.8 # 0.5
end

scen4a = Static.@scenario Static.ExoAlpha begin
    m = 0.7 # 0.15
    γ = 0.8 # 0.5
end
```

```julia
baseline=solve_model(Static.ExoAlpha)
exp1=solve_model(scen1a)
exp2=solve_model(scen2a)
exp3=solve_model(scen3a)
exp4=solve_model(scen4a)
```

```julia
display(html"<h3>Baseline</h3>")
display(baseline.sheets)
display(html"<h3>Exogenous Alpha</h3>")
display(exp1.sheets)
display(html"<h3>Central Bank Sense</h3>")
display(exp2.sheets)
display(html"<h3>Dual Interest</h3>")
display(exp3.sheets)
```

```julia
Static.build_table_from_solutions([baseline, exp1, exp2, exp3, exp4])
```

## Evaluating curves

The code for evaluating curves is given as follows: 
The first arguement of the function is the model whose curves should be evaluated. The second argument is a dictionary with the variable names as keys and the values at which the curves should be evaluated as values. The function returns a named tuple of curves evaluated at that point

```julia
Static.eval_curve(Static.AssetPK, solve_model(Static.AssetPK).variables)
```

Example IR/IS Curve in $Y$

```julia
IS(r) = begin 
    sol = solve_model(Static.AssetPK).variables
    sol[:r] = r
    Static.eval_curve(Static.AssetPK, sol).IS
end
```

```julia
IR(Y) = begin 
    sol = solve_model(Static.AssetPK).variables
    sol[:Y] = Y
    Static.eval_curve(Static.AssetPK, sol).IR
end
```

### Example Plot

```julia
f = Figure()
a = Axis(f[1, 1], title = "IS curve", xlabel = "r", ylabel = "Y")
b = Axis(f[1, 2], title = "IR curve", xlabel = "Y", ylabel = "r")
lines!(a, [IS(x) for x in 0:0.01:1])
lines!(b, [IR(x) for x in 0:10:100])
```
