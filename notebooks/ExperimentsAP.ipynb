```julia

using PKAssetPrices.Dynamic: DynamicModel, DiscreteTime, DynamicParametrization, solve_model, @model
using PKAssetPrices
using CairoMakie
```

```julia

sol_Q1 =  solve_model(Dynamic.DynQconAPLevelChange)
sol_Q2 =  solve_model(Dynamic.DynQconAPChange)
sol_Q3 =  solve_model(Dynamic.DynQoldAPLevelChange)
sol_Q4 =  solve_model(Dynamic.DynQoldAPChange)
```
