```julia

using PKAssetPrices.Dynamic: DynamicModel, DiscreteTime, DynamicParametrization, solve_model, @model
using PKAssetPrices
using CairoMakie
```

```julia

sol_Q1 =  solve_model(Dynamic.DynQconAPLevelChange)
sol_Q2 =  solve_model(Dynamic.DynQconAPChange)
sol_Q3 =  solve_model(Dynamic.DynQoldAPLevelChange)
sol_Q4 =  solve_model(Dynamic.DynQoldAPChange)
```

```julia
f = Figure()
ax = Axis(f[1,1])
lines!(ax, sol_Q2.paths[:AD], label="Contemporary AP Change only")
lines!(ax, sol_Q3.paths[:AD], label="Old AP Level & change ")
lines!(ax, sol_Q4.paths[:AD], label="Old AP Change only")
axislegend(ax)
f
```

```julia
f = Figure()
ax = Axis(f[1,1])
lines!(ax,sol_Q2.paths[:AP], label="Contemporary AP Change only")
lines!(ax,sol_Q3.paths[:AP], label="Old AP Level & change ")     
lines!(ax,sol_Q4.paths[:AP], label="Old AP Change only")         
axislegend(ax)
f
```
